## 1. Numbers divisible by 7 but not by 3

Find all numbers from 2000 to 3200 (inclusive) that are divisible by 7 but are not multiples of 3. Print them as a comma-separated sequence.

```python
numbers = []

for i in range(2000, 3201):
    if i % 7 == 0 and i % 3 != 0:
        numbers.append(str(i))

print(",".join(numbers))
```

## 2. Convert a list of characters into a string

Example input: `['C', 'O', 'D', 'I', 'N', 'G']`  
Expected output: `CODING`

```python
characters = ['C', 'O', 'D', 'I', 'N', 'G']

result = "".join(characters)

print(result)
```

## 3. Create a dictionary of numbers and their squares

Given an integer `n`, create a dictionary containing each number from 1 to `n` and its square.

```python
n = int(input("Enter a number: "))

squares = {}

for i in range(1, n + 1):
    squares[i] = i * i

print(squares)
```
